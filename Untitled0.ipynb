count no of vowels and consonents in strings

```python
a=input("enter the string: ")
b=a.lower()
v=b.count('a')+a.count('e')+a.count('i')+a.count('o')+a.count('u')
print("vowels :",v)
l=len(b)
c=l-v
print("consonants:",c)
```
