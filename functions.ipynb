```python
def palindrome(t: str):
  t=t.lower()
  if(t==t[::-1]):
    print("palindrome")
  else:
    print("not a palindrome")
t=input("enter a string")
palindrome(t)
```

program that reverse given string using functions

```python
def rev(t: str):
  k=t[::-1]
  return k
rev(input("enter string"))
```

```python
def reverse(st):
  rev=""
  for i in range(len(st)-1,-1,-1):
    rev+=st[i]
  return rev
st=input()
print(reverse(st))
```

function that returns multiple values

```python
def multi(*num):
  s=__builtins__.sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
total_sum,count,avg=multi(1,2,3,4,5)
print(f"sum is{total_sum},count is {count},average is {avg}")
```
